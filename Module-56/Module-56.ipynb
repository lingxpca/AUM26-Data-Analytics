```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
```

```python
ht = pd.read_csv("https://raw.githubusercontent.com/lingxpca/AUM26-Data-Analytics/refs/heads/main/Module-56/mother_daughter.csv")
```

```python
plt.figure(figsize=(5,5))
plt.scatter(ht['mother_stature'],ht['daughter_stature'],s=1)
```

```python
md = sm.OLS(ht['daughter_stature'],sm.add_constant(ht['mother_stature'])).fit()
md.summary()
```

```python
pred = md.predict(sm.add_constant(ht['mother_stature']))
np.sum((pred - ht['daughter_stature'])**2) #total squared residual
```

```python
abalone = fetch_ucirepo(id=1)
```

```python
X = abalone.data.features
y = abalone.data.targets
df = pd.concat([X,y],axis=1)
df['old'] = (df['Rings']> 9).astype(int)
```

```python
abalone_logit = smf.logit('old~C(Sex)+Length+Diameter+Height+Whole_weight',
                          data=df).fit()
abalone_logit.summary()
np.exp(abalone_logit.params) #odds ratio
#for weight: every one unit increase on whole_weight, the odds of being older is increase by 59%; holding other var fixed
#for length: every one unit increase on length, the odds of being older is decrease by 1-2.034762e-04,holding other var fixed
#for sex; compared to female, the odds of being older is decreased by 1-0.4=60% for infant abalone
#fro sex; compared to female, the odds of being older is decreased by 1-0.9=10% for male abalonen
```

```python
from sklearn.metrics import accuracy_score
from sklearn.datasets import fetch_lfw_people
from sklearn.neural_network import MLPClassifier
```

```python
lp = fetch_lfw_people(min_faces_per_person=100,resize=0.4)
```

```python
X=lp.data
y=lp.target
lp.target_names
```

```python
n_samples,h,w=lp.images.shape
print(n_samples,h,w)
```

```python
plt.figure(figsize=(10,5))
for i in range(10):
    plt.subplot(2,5,i+1)
    plt.imshow(lp.images[i],cmap='gray')
    plt.title(lp.target_names[lp.target[i]])
```

```python
#original 1850 -> 128 -> 64 -> 5
nn = MLPClassifier(
  hidden_layer_sizes=(128),
  activation="relu",
)
nn.fit(X,y)
```

```python
y_pred = nn.predict(X)

plt.figure(figsize=(15,10))
for i in range(10):
    plt.subplot(2,5,i+1)
    plt.imshow(lp.images[i],cmap='gray')
    actual=lp.target_names[lp.target[i]]
    predicted=lp.target_names[y_pred[i]]
    plt.title(f"Actual: {actual}\nPredicted: {predicted}")
```

```python
accuracy_score(y,y_pred)
```
